### **Función SUMAR**

```python
%%writefile suma.py
def sumar(a, b):
  return a + b
```

```python
# Prueba pytest para función SUMA:
%%writefile test_sumar.py
from suma import sumar

def test_sumar_positivos():
  assert sumar(2, 3) == 5

def test_sumar_negativos():
  assert sumar(-1, -4) == -5
```
